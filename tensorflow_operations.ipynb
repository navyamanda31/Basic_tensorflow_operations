```python
import tensorflow as tf
scalar = tf.constant(10)
print(scalar)
```

```python
vector = tf.constant([10,25,30])
print(vector)
```

```python
matrix = tf.constant([
    [1,2],
    [3,4]
])
print(matrix)
```

```python
tensor_3d = tf.constant([[[1, 2],[3,4]],[[5,6],[7,8]]])

print(tensor_3d)
```

```python
a = tf.constant([10,20])
b = tf.constant([30,40])

print(tf.add(a,b))
```

```python
print(tf.subtract(a,b))
```

```python
print(tf.multiply(a,b))
```

```python
matrix1 = tf.constant([[1,2],[3,4]])
matrix2 = tf.constant([[5,6],[7,8]])
print(tf.matmul(matrix1,matrix2))
```

```python
tensor = tf.constant([1,2,3,4,5,6])
reshaped = tf.reshape(tensor,(2,3))
print(reshaped)
```
